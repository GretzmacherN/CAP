Loading Samples

```python
import os
import re
import pandas as pd
import numpy as np
import scanpy as sc
from collections import defaultdict

def get_condition(fname: str) -> str:
    if "HER2" in fname:                          return "HER2"
    if "-ER-"  in fname or "_ER-"  in fname:     return "ER"
    if "-TN-"  in fname or "_TN-"  in fname:     return "TN"
    if "-N-"  in fname or "_N-"  in fname: return "N"
    return "unknown"
```

```python
#loading files
data_dir = "/Volumes/TOSHIBA EXT/Nowotny/ENV1/head"

adatas          = []#keeping all AnnData objects in memory
summary_list    = []#per-sample stats
condition_hvgs  = defaultdict(lambda: None)#HVG intersection per condition
skipped         = []
```

```python
for fname in sorted(os.listdir(data_dir)):
    if not fname.endswith(".h5ad") or fname.startswith("._"):
        continue

    path      = os.path.join(data_dir, fname)
    condition = get_condition(fname)

    try:
        adata = sc.read_h5ad(path)
    except Exception as e:
        skipped.append(f"{fname}: ERROR – {e}")
        continue

    #Summary statistics
    full_patient = adata.obs["patient_id"].iloc[0]
    #Removing GSM prefix if present
    patient = re.sub(r"^GSM\d+_", "", full_patient)
    #Extracting condition
    match = re.match(r"^(HER2|TN|ER|N)", patient)

    if match:
        condition_code = match.group(1)
    else:
        condition_code = "UNKNOWN"

    print(condition_code)
    summary_list.append({
        "patient_id":   patient,
        "condition":    condition_code,
        "n_cells":      adata.n_obs,
        "mean_counts":  adata.obs["total_counts"].mean(),
        "var_counts":   adata.obs["total_counts"].var(),
    })

    #HVG intersection
    if "highly_variable" in adata.var.columns:
        hvgs = set(adata.var_names[adata.var["highly_variable"]].tolist())
        condition_hvgs[condition] = (
            hvgs if condition_hvgs[condition] is None
            else condition_hvgs[condition] & hvgs
        )
        print(f"Loaded: {fname}  [{condition}]  →  {adata.shape},  {len(hvgs)} HVGs")
    else:
        skipped.append(f"{fname}: no highly_variable column")
        print(f"Loaded: {fname}  [{condition}]  →  {adata.shape}  (no HVGs)")

    adatas.append(adata)
```

```python
print(f"Processed samples: {len(summary_list)}")
print(f"Skipped samples: {len(skipped)}")

if len(summary_list):
    print(pd.DataFrame(summary_list).columns)
else:
    print("summary_list is empty")

print(skipped[:10])
```

```python
#summary DataFrame
summary_df = (
    pd.DataFrame(summary_list)
    .sort_values(["condition", "patient_id"])
    .reset_index(drop=True)
)
```

```python
#HVG report
print(f"\nLoaded {len(adatas)} samples,  {len(skipped)} skipped")
if skipped:
    print("\nSkipped:")
    for s in skipped: print(f"  {s}")

print("\n=== HVG intersection per condition ===")
final_hvgs_union        = set()
final_hvgs_intersection = None

for cond, hvgs in condition_hvgs.items():
    if hvgs is None:
        print(f"  {cond}: no HVGs found")
        continue
    print(f"  {cond}: {len(hvgs)} HVGs")
    final_hvgs_union.update(hvgs)
    final_hvgs_intersection = (
        hvgs if final_hvgs_intersection is None
        else final_hvgs_intersection & hvgs
    )

print(f"\nFinal gene set options:")
print(f"  Union across conditions:        {len(final_hvgs_union)} genes")
print(f"  Intersection across conditions: {len(final_hvgs_intersection or set())} genes")
```

```python
epithelial_types = [
    "Epithelial",
    "Epithelial: Basal",
    "Epithelial: Luminal Progenitor",
    "Epithelial: Mature Luminal"]

epi_mask = adata.obs["cl_annotation"].isin(epithelial_types)
adata_epi = adata[epi_mask].copy()

median_cnv = adata_epi.obs["cnv_score"].median()
adata_epi.obs["cnv_group"] = np.where(
    adata_epi.obs["cnv_score"] >= median_cnv, "High CNV", "Low CNV"
)

print(f"CNV threshold (median): {median_cnv:.4f}")
print(adata_epi.obs["cnv_group"].value_counts())

conditions = adata_epi.obs["condition"].unique()
MAX_CELLS_PER_GROUP = 2000

for cond in sorted(conditions):
    print(f"\nPlotting condition: {cond}")

    cond_mask = adata_epi.obs["condition"] == cond
    if cond_mask.sum() == 0:
        print(f"  No cells for condition {cond}, skipping.")
        continue

    cond_idx = np.where(cond_mask)[0]
    X_cnv = adata_epi.obsm["X_cnv"][cond_idx]
    if sp.issparse(X_cnv):
        X_cnv = X_cnv.toarray()
    X_cnv = X_cnv.astype(np.float32)

    obs = adata_epi.obs.iloc[cond_idx].copy()
    obs["_idx"] = np.arange(len(obs))

    obs_sorted = obs.sort_values(
        ["cnv_group", "cnv_score"], ascending=[True, False]
    )

    # Downsample per group, then re-sort to preserve High CNV first
    groups = []
    for grp, grp_df in obs_sorted.groupby("cnv_group", sort=False):
        if len(grp_df) > MAX_CELLS_PER_GROUP:
            grp_df = grp_df.sample(n=MAX_CELLS_PER_GROUP, random_state=42)
        groups.append(grp_df)
    obs_sorted = pd.concat(groups).sort_values(
        ["cnv_group", "cnv_score"], ascending=[True, False]
    )

    sorted_idx = obs_sorted["_idx"].values
    X_sorted = X_cnv[sorted_idx, :]

    # Build group boundary lines and labels
    group_labels = obs_sorted["cnv_group"].values
    boundaries = []
    label_positions = []
    label_names = []
    current = group_labels[0]
    start = 0

    for i, g in enumerate(group_labels):
        if g != current:
            boundaries.append(i)
            label_positions.append((start + i) / 2)
            label_names.append(current)
            current = g
            start = i

    label_positions.append((start + len(group_labels)) / 2)
    label_names.append(current)

    # Plot — only ONE fig/ax creation
    fig, ax = plt.subplots(figsize=(20, max(6, len(obs_sorted) / 80)))

    im = ax.imshow(X_sorted, aspect="auto", cmap="RdBu_r", vmin=-0.5, vmax=0.5, interpolation="none")

    for b in boundaries:
        ax.axhline(b - 0.5, color="black", linewidth=1.5, linestyle="--")

    ax.set_yticks(label_positions)
    ax.set_yticklabels(label_names, fontsize=12, fontweight="bold")

    # Chromosome boundaries — use adata_epi.uns, not adata_cond
    if "cnv" in adata_epi.uns and "chr_pos" in adata_epi.uns["cnv"]:
        chr_pos = adata_epi.uns["cnv"]["chr_pos"]
        for chrom, pos in chr_pos.items():
            ax.axvline(pos - 0.5, color="black", linewidth=0.5, alpha=0.5)
            ax.text(
                pos, -0.02, chrom.replace("chr", ""),
                transform=ax.get_xaxis_transform(),
                fontsize=7, ha="center", va="top", rotation=0
            )

    ax.set_xlabel("Genomic position (by chromosome)", fontsize=12)
    ax.set_title(f"InferCNV — {cond}", fontsize=14, fontweight="bold")

    cbar = plt.colorbar(im, ax=ax, shrink=0.4, pad=0.01)
    cbar.set_label("CNV signal", fontsize=10)
    cbar.ax.tick_params(labelsize=8)

    plt.tight_layout()
    plt.savefig(rf"{data_dir}/../Figures/12_CNV_heatmap_{cond}.png", dpi=150, bbox_inches="tight")
    plt.close(fig)  # free memory — don't plt.show() inside the loop
    print(f"  Saved: CNV_heatmap_{cond}.png")
```

```python
import anndata as ad
import scipy.sparse as sp

#Global reference populations — Sample 9260 (all cell types) as reference
ref_mask = (adata.obs["sample"] == "N-N1105-Epi")
adata_ref = adata[ref_mask].copy()
print(f"Global references (9260, all cell types): {adata_ref.n_obs}")

#Storage
adata.obs["cnv_score"] = np.nan
adata.obs["cnv_signal"] = np.nan

#running sample-wise inferCNV
samples = adata.obs["sample"].unique()

for sample in samples:

    print("\n----------------------------")
    print(f"Processing: {sample}")

    #skipping reference sample
    if sample == "N-N1105-Epi":
        print("Skipping 9260 (is the reference sample)")
        continue

    #targetig sample
    target = adata[adata.obs["sample"] == sample].copy()
    global_refs = adata_ref.copy()

    ad_sub = ad.concat([target, global_refs], merge="same")
    ad_sub.obs_names_make_unique()

    ad_sub.X = ad_sub.layers["counts"].copy()#Raw counts

    if not sp.issparse(ad_sub.X):
        ad_sub.X = sp.csr_matrix(ad_sub.X)

    #Reference annotation
    ad_sub.obs["cnv_reference"] = "other"

    #All cells from sample 9260 are reference
    ref_cells = ad_sub.obs["sample"] == "N-N1105-Epi"
    ad_sub.obs.loc[ref_cells, "cnv_reference"] = "reference_9260"

    #inferCNV
    try:

        cnv.tl.infercnv(
            ad_sub,
            reference_key="cnv_reference",
            reference_cat=["reference_9260"],
            window_size=100,
            step=10,
            dynamic_threshold=0.5,
            exclude_chromosomes=("chrX", "chrY"),
            n_jobs=1
        )

        #copying chromosome metadata back to main adata only once
        if "cnv" not in adata.uns:
            adata.uns["cnv"] = ad_sub.uns["cnv"].copy()

        X_cnv = ad_sub.obsm["X_cnv"]

        #dense -> sparse
        if sp.issparse(X_cnv):
            X_cnv = X_cnv.toarray()

        #computing scores
        cnv_score = np.mean(np.abs(X_cnv), axis=1)
        cnv_signal = np.max(np.abs(X_cnv), axis=1)

        #assigning to ad_sub.obs
        ad_sub.obs["cnv_score"] = cnv_score
        ad_sub.obs["cnv_signal"] = cnv_signal

        #defining target cells and writing back
        target_mask = ad_sub.obs["sample"] == sample
        target_names = ad_sub.obs_names[target_mask]

        adata.obs.loc[target_names, "cnv_score"] = ad_sub.obs.loc[target_names, "cnv_score"]
        adata.obs.loc[target_names, "cnv_signal"] = ad_sub.obs.loc[target_names, "cnv_signal"]

        if "X_cnv" not in adata.obsm:
            adata.obsm["X_cnv"] = np.zeros(
                (adata.n_obs, X_cnv.shape[1])
            )

        target_idx = adata.obs_names.get_indexer(target_names)
        adata.obsm["X_cnv"][target_idx, :] = X_cnv[
            ad_sub.obs_names.get_indexer(target_names), :
        ]

        print("Success")
        print(ad_sub.obs["cnv_score"].describe())

    except Exception as e:

        print(f"Failed on {sample}")
        print(e)

    del target
    del global_refs
    del ad_sub

    gc.collect()
```
